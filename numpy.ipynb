```python
import numpy as np
array1=np.array([1,2,3])
array2=np.array([2,3,4])
print(array1+array2) # list ni har bir elementini bir biriga qo'shadi
```

```python
scores=np.array([40,60,30,99,76])
print(scores>=60) # yani bittalab aylanib chiqilmaydi shunchaki shart berilsa o'zi aylanib chiqadi
```

```python
scores=np.array([40,60,30,99,76])
scores[scores>=60]=0 # ichidagi shart listni aylanib true falce qiymat qaytaradi va 60 dan katta qiymat chiqsa  arrayning shu sonni 0 change qil
print(scores)
```

```python
array1=np.array([[1,2,3,4]])
array2=np.array([[5],[6],[7],[8]])
print(array1.shape)
print(array2.shape)
print(array1*array2) # shart shuki ularning column laridan biri yoki 1 bo'lishi yoki ikkalasi teng bo'lishi kk va row uchun ham yoki 1 tasi 1 ga teng bo'lishi kk or bir biriga teng bo'lishi kk
```

```python
array=np.array([[1,2,3,4],
                [5,6,7,8]])
print(np.sum(array, axis=0)) # axis=0 da 2 ta array ni column larini qo'shib beryapti
print(np.sum(array, axis=1)) # axis=0 da 2 ta array ni row larini qo'shib beryapti
```

```python
array=np.array([[13,21,38,49],
                [55,16,27,18]])
teenagers=array[array<=18] # bunda pythond kabi for loop shart emas umuman
print(teenagers,"\n")

even=array[array%2==0] # array ni ichidan even numberslarni chiqaraib olish
print(even,"\n")
 # bunda bizni 2d arrayimizni 1d ga convert qip qo'yayapti agar biz 2d ni saqlab  qolmoqchi bo'lsak quyidagicha yozamiz
array1=np.where(array>=18,array,0) # here we use np.where and replace the numbers under 18 with 0
print(array1)
```

```python
import numpy as np

rng = np.random.default_rng()

fruits = np.array(["apple", "orange", "banana", "coconut", "pineapple"])
fruits = rng.choice(fruits, size=(3, 3)) # randomley tanlayapti
print(fruits)
```

```python
import numpy as np

rng = np.random.default_rng()

numbers = np.array([1, 2, 3,4, 5])
rng.shuffle(numbers) # array ni ichidagi sonlarni randomli aralashitiryapti
print(numbers)
```
