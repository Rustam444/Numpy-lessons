```python
import time
start_time = time.time() # label start time
prices = np.array([120, 85, 45, 210, 60, 150, 35, 180])
for index, values in enumerate(prices):
  if values > 100:
    prices[index]=values*0.15
print(prices)
print(time.time()-start_time) # final time - start time
```

```python
start_time=time.time()
prices = np.array([120, 85, 45, 210, 60, 150, 35, 180])
prices[prices>100]=prices[prices>100]*0.15 # say that if the elements of prices > 100, multiply 0.15 each of them and put it in that place
print(prices)
print(time.time()-start_time)
```
