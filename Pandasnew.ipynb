```python
# in pandas we have loc() and iloc() that helps us to find location.
import pandas as pd
array=["salom", "karam","olma","anor"]

new=pd.Series(array, index=["a","b","c","d"]) # it gives us in order this array like excel

print(new.iloc[3]) # we use iloc[]- which means that integer location it helps us to find the word located in 3 index, we also find any elements through their indexes

print(new.loc["a"])# here just loc[]-location as you can see we gave that array indexes which are "a" "b" ... so can get them with "a" or other indexes

# so we also change their value through their index
print()
new.loc["a"]="salom dunyo" #
print(new)
```

```python
# so it is similar for dict
calories = {"Day 1": 1750, "Day 2": 2100, "Day 3": 1700}

print(calories["Day 1"])
print(calories["Day 2"],"\n")

calories["Day 1"]=200 # here we change Day 1's value
print(calories)
```

```python
calories = {"Day 1": [1750,1200], "Day 2": [2100,300], "Day 3": [1700]}
df=pd.DataFrame(calories)
print(df)
```

```python
data = {
"Name": ["Spongebob", "Patrick", "Squidward"], # bu bizga berilgan dict edi
"Age": [30, 35, 50]
}

df=pd.DataFrame(data)

df["Name"]=["Rustam","Alisher","Elbek"] # buni xuddi dict qoidasidek key qiymatiga yangi value berib dict ni yangilayapmiz
df["Age"]=[21,23,30] # Age qismida ham shunday as you can see, farqi value qismini ham list da berish kk

print(df,"\n")

df["Job"]=["ML engeener","Arxitektor","Software engeener"] # as you can see job keys was not there, but now we have with its new values
print(df)
```

```python
# Add a new row
new_row = pd.DataFrame ({"Name": "Sandy", "Age": 28, "Job": "Engineer"}, index=[4])
df = pd.concat([df, new_row])
print(df)
```
